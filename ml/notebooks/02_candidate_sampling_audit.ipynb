# Candidate Sampling Audit (Batch 001)

This notebook analyzes the event population and executes the candidate sampling workflow for human labeling.

```python
import pandas as pd
import sys
from pathlib import Path
sys.path.append(str(Path('.').resolve().parent))
from src.data_loader import load_v2_features
from src.feature_schema import FeatureSchemaValidator
from src.candidate_sampler import CandidateSampler

FEATURES_PATH = "../data/processed/features/event_features_v2.parquet"
FEATURE_SCHEMA = "configs/feature_schema.json"
OUT_CSV = "data/candidate_events_batch_001.csv"
OUT_PARQUET = "data/candidate_events_batch_001_features.parquet"
```

## 1. Load Data & Validate

```python
df = load_v2_features(FEATURES_PATH)
validator = FeatureSchemaValidator(FEATURE_SCHEMA)
sampler = CandidateSampler(df, validator, seed=42)

print(f"Loaded {len(df)} events from source.")
```

## 2. Execute Stratified Sampling

```python
candidates = sampler.generate_batch(Path(OUT_CSV), Path(OUT_PARQUET), n_target=400)
print(f"\nSampled {len(candidates)} events.")
```
