# Labeling Audit Notebook

Analyzes the progress and inter-rater agreement of human labeling for Batch 001.

```python
import pandas as pd
import numpy as np

WORKSHEET_PATH = "data/batch_001_reviewer_worksheet.csv"
try:
    df = pd.read_csv(WORKSHEET_PATH)
except FileNotFoundError:
    df = pd.DataFrame()

if not df.empty:
    print(f"Loaded worksheet with {len(df)} candidate events.")
    print("\n--- Reviewer 1 Progress ---")
    print(f"Completed: {df['reviewer_1'].notna().sum()}")
    print("\n--- Reviewer 2 Progress ---")
    print(f"Completed: {df['reviewer_2'].notna().sum()}")
    print("\n--- Final Consensus Progress ---")
    print(f"Completed: {df['final_label'].notna().sum()}")
```
